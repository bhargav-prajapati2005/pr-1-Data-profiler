#### Q2. How to Plan a Data Science Project?

##### The following steps are used to plan a Data Science project:

###### Problem Definition: Understand the problem that needs to be solved.

###### Data Collection: Collect data from CSV, JSON, databases, and APIs.

###### Data Understanding: Study the structure and characteristics of the data.

###### Data Cleaning: Handle missing values, duplicate records, and incorrect data.

###### Exploratory Data Analysis: Analyze data using graphs and statistics.

###### Feature Engineering: Create useful features from existing data.

###### Data Preprocessing: Encode categorical data and scale numerical values.

###### Model Building: Train a suitable Machine Learning model.

###### Model Evaluation: Check model performance.

###### Deployment: Make the model available for practical use.

###### Monitoring: Monitor model performance and update it when required.

#### Q4. What are Tensors?
A tensor is a mathematical structure used to store numerical data in multiple dimensions.

Tensors are widely used in Machine Learning and Deep Learning

#### Types of Tensors

###### 0D Tensor – Scalar

A single numerical value.

```python
import numpy as np
a = np.array(10)
print(a)
```

##### 1D Tensor – Vector

A collection of values.

```python
a = np.array([10, 20, 30])
print(a)
```

###### 2D Tensor – Matrix

Data arranged in rows and columns.

```python
a = np.array([[1, 2],
              [3, 4]])
print(a)
```

#### 3D Tensor

A collection of multiple matrices.

```python
a = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print(a.shape)
```

Important Tensor Properties:
- ndim: Returns the number of dimensions.
- shape: Returns the size of each dimension.
- dtype: Returns the data type.
- size: Returns the total number of elements.

Example:

Example:


```python
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])

print("Dimensions:", a.ndim)
print("Shape:", a.shape)
print("Data Type:", a.dtype)
print("Size:", a.size)
```

Tensors are useful for storing and processing large amounts of numerical data in Machine Learning and Neural Networks.
